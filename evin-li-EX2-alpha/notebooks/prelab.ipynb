# EX2: Alpha Particle Range in Air 

## Alpha Prelab

Simulating passage of alpha particles through air. We do so by defining an interaction policy for a single thin layer of air and repeating for N layers. The policy is as follows: particles can either interact once (which causes it to lose some amount of energy $dE$) OR pass right through. The input of this simulated column of air is a population of particles with a Gaussian distribution of energies centered at $E0$ and some spread $\sigma$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
E0 = 5.0      # Initial midpoint of Gaussian distribution
sigma = 0.1   # Initial spread/S.D.
minE = 0.3    # Lower bound of E, considered stopped below this point
F = 0.05      # 10% chance of interaction per layer
dE = 0.1      # Energy lost per interaction
```

```python
# Plot initial distribution of energies
particles = np.random.normal(E0, sigma, 10000)

plt.hist(particles, bins=50)
plt.title("Initial Alpha Energy Distribution")
plt.ylabel("Number of Particles")
plt.xlabel("Energy (MeV)")
plt.show()
```

```python
# Plot initial distribution of energies (with fixed x-axis range)
particles = np.random.normal(E0, sigma, 10000)

plt.hist(particles, bins=50)
plt.title("Initial Alpha Energy Distribution (Fixed X Scale)")
plt.ylabel("Number of Particles")
plt.xlabel("Energy (MeV)")
plt.xlim(0, 5 + 3 * sigma) # 3 SD = 99.7% of data
plt.show()
```

```python
# Reset particle energies
particles = np.random.normal(E0, sigma, 10000)
# Plot energies after 100 layers
layers = 100

for layer in range(layers):
    mask = particles > minE  # Only process active particles
    interactions = np.random.random(len(particles[mask])) < F
    particles[mask] -= interactions * dE

# Plotting
plt.hist(particles[particles > minE], bins=50)
plt.title(f"Alpha Energy Distribution after {layers} Layers")
plt.ylabel("Number of Particles")
plt.xlabel("Energy (MeV)")
plt.xlim(0, 5 + 3 * sigma) # 3 SD = 99.7% of data
plt.show()
```
